# CUDA on Colab — Matrix Multiplication

**Before you run anything:** menu **Runtime → Change runtime type → T4 GPU → Save**.

Then menu **Runtime → Run all** (or press the ▶ button on the cell below). It will:
1. confirm the GPU,
2. download the code,
3. build it for the T4,
4. run it at n=5000 and n=10000,
5. download the two result files.

It takes under a minute. Free Colab is enough — no payment needed.

```python
# === CUDA on Colab — runs everything at once ===

# 1) confirm the GPU (should say Tesla T4)
!nvidia-smi -L

# 2) get the code (safe to re-run)
![ -d HPC-benchmark ] || git clone https://github.com/MaryamHajishah/HPC-benchmark.git
%cd /content/HPC-benchmark

# 3) build for the T4 (compute capability 7.5)
!nvcc -O3 -arch=sm_75 cuda/matmul_cuda.cu -o matmul_cuda -lcublas

# 4) run and save the output
!mkdir -p outputs
!./matmul_cuda 5000  2>&1 | tee outputs/cuda_5000.txt
!./matmul_cuda 10000 2>&1 | tee outputs/cuda_10000.txt
# !./matmul_cuda 15000 2>&1 | tee outputs/cuda_15000.txt   # optional 3rd size (remove the # to use)

# 5) download the result files to your computer
from google.colab import files
files.download('outputs/cuda_5000.txt')
files.download('outputs/cuda_10000.txt')
```
